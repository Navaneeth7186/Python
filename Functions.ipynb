```python
#Grade Calulation
def find_grade(mark):
    if mark >= 90:
        return "A"
    elif mark >= 80:
        return "B"
    elif mark >= 70:
        return "C"
    elif mark >= 60:
        return "D"
    elif mark >= 40:
        return "E"
    else:
        return "Fail"
mark = int(input("Enter marks:"))

print("Grade =", find_grade(mark))
    
```
